# Modelo de Incêndio Probabilístico (Fire Model Prob)
### Dinâmica de fogo com combustão espontânea e regeneração

Este modelo expande o modelo de incêndio básico ao introduzir processos estocásticos (aleatórios). Em vez de apenas propagar o fogo, o sistema agora permite que novas chamas surjam espontaneamente e que áreas queimadas se recuperem ao longo do tempo.

## Conceitos Fundamentais

O modelo utiliza três estados:
- **Floresta** (0): Cor verde.
- **Em chamas** (1): Cor vermelha.
- **Queimado** (2): Cor cinza.

**Regras Probabilísticas (Vizinhança de Queen/8 direções):**

1. **Combustão Espontânea:** Uma célula de **Floresta** pode pegar fogo se um vizinho estiver queimando OU com uma pequena probabilidade $p_{combustao}$ (simulando raios ou causas externas).
2. **Extinção:** Células **Em chamas** sempre se tornam **Queimadas** no próximo passo.
3. **Regeneração:** Uma célula **Queimada** tem uma probabilidade $p_{regeneracao}$ de voltar a ser **Floresta** a cada passo.

Estas regras criam um ciclo contínuo onde a floresta nunca desaparece completamente, estabelecendo um equilíbrio dinâmico.

## Imports

````python
from matplotlib.colors import ListedColormap

from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import FireModelProb
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  FireModelProb  →  Map
```

````python
env = Environment(start_time=0, end_time=50)
````

## Construção do espaço/modelo

Criamos uma grade de 30x30.

````python
gdf = vector_grid(dimension=(30, 30), resolution=1, attrs={"state": 0})
gdf.head()
````

## Inicialização

Configuramos as probabilidades de combustão e regeneração. Valores típicos são pequenos para a combustão e moderados para a regeneração.

````python
model = FireModelProb(gdf=gdf)
model.setup(
    prob_combustion=0.005, 
    prob_regrowth=0.05
)
# Inicializamos com tudo como Floresta (estado 0)
gdf["state"] = 0
````

## Visualização

Usaremos o mesmo esquema de cores do modelo básico.

````python
cmap = ListedColormap(["#27ae60", "#e74c3c", "#7f8c8d"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "none"},
)
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Muitos Raios:** Aumente `prob_combustion` para 0.05. A floresta consegue se recuperar a tempo ou vive em constante queima?

O modelo probabilístico revela como sistemas naturais mantêm a resiliência através da regeneração. A inclusão de eventos aleatórios permite a simulação de cenários mais realistas de manejo florestal e ecologia de paisagens.
