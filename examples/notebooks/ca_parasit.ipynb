# Modelo Hospedeiro-Parasita (Parasit)
### Dinâmica espacial e coexistência em ondas

Este modelo, baseado em Hassell et al. (1991), simula a interação espacial entre uma população de hospedeiros (como insetos) e seus parasitas. O modelo é famoso por gerar padrões espaciais complexos que permitem a coexistência estável das duas populações.

## Introdução

Na ecologia, interações hospedeiro-parasita muitas vezes resultam em oscilações populacionais. Quando o espaço é incluído na simulação, essas oscilações se tornam ondas que se propagam, colidem e se aniquilam, criando um mosaico dinâmico. O comportamento emergente é o caos espacialmente organizado.

## Conceitos Fundamentais

### Ciclo de Vida (Estados)
As células passam por **9 estados (0 a 8)** que representam as fases combinadas:
- **Estado 0**: Hospedeiro suscetível.
- **Estados 1-2**: Hospedeiro infectado (avança automaticamente).
- **Estado 3**: Parasita latente.
- **Estados 4-8**: Ciclo do parasita (amadurecimento e morte, avança automaticamente até voltar ao estado 0).

### Regras de Transição
1. Um hospedeiro suscetível (**0**) torna-se infectado (**1**) se algum vizinho for um hospedeiro infeccioso (**1**).
2. Um parasita latente (**3**) torna-se ativo (**4**) se algum vizinho for um parasita maduro (**5**).
3. Outros estados avançam um passo por vez no ciclo de vida.

**Vizinhança**: Von Neumann (4 vizinhos - Rook).

## Imports

````python
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Parasit
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  Parasit  →  Map
```

````python
env = Environment(start_time=0, end_time=300)
````

## Construção do espaço/modelo

Criamos uma grade de 50x50.

````python
gdf = vector_grid(dimension=(50, 50), resolution=1, attrs={"state": 0})
````

## Inicialização

O modelo começa com uma distribuição aleatória de todos os estados, simulando uma comunidade mista inicial.

````python
model = Parasit(gdf=gdf)
model.initialize()
````

## Visualização

````python
Map(gdf=gdf, plot_params={"column": "state", "cmap": "tab10", "ec": "none"})
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

## Conclusão

O modelo Hospedeiro-Parasita demonstra como a estrutura espacial é crucial para a ecologia. Sem o espaço, uma população poderia extinguir a outra; com o espaço, frentes de onda de infecção e recuperação permitem a coexistência infinita.
