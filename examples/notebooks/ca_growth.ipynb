# Modelo de Crescimento (Growth)
### Expansão espacial a partir de uma semente inicial

Este modelo simula um processo de colonização ou crescimento estocástico, onde uma semente inicial se expande para células vizinhas com uma determinada probabilidade.

## Introdução

Modelos de crescimento são fundamentais para entender fenômenos como a expansão de cidades, a propagação de epidemias ou a colonização de uma espécie em um novo território. O comportamento emergente aqui é a formação de um aglomerado (cluster) cujo formato e densidade dependem da probabilidade de colonização.

## Conceitos Fundamentais

### Estados
- **Vazio (0)**: Célula disponível para colonização.
- **Vivo (1)**: Célula colonizada (semente ou descendente).

### Regras de Transição
1. Uma célula **Viva** permanece Viva para sempre (não há morte).
2. Uma célula **Vazia** torna-se **Viva** com probabilidade $P$ se tiver pelo menos um vizinho Vivo.

**Vizinhança**: Queen (8 vizinhos).

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Growth
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  Growth  →  Map
```

````python
env = Environment(start_time=0, end_time=30)
````

## Construção do espaço/modelo

Criamos uma grade vetorial de 30x30.

````python
gdf = vector_grid(dimension=(30, 30), resolution=1, attrs={"state": 0})
````

## Inicialização

O modelo de crescimento coloca automaticamente uma semente no centro da grade ao chamar `initialize()`.

````python
# Definimos a probabilidade de crescimento (ex: 20%)
model = Growth(gdf=gdf, dim=30, probability=0.2)
model.initialize()
````

## Visualização

````python
cmap = ListedColormap(["#f0f0f0", "forestgreen"])
Map(gdf=gdf, plot_params={"column": "state", "cmap": cmap, "ec": "#ccc"})
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Probabilidade**: O que acontece se a probabilidade for muito baixa (ex: 0.05) ou muito alta (ex: 0.8)? Como isso afeta a forma do cluster?
- **Tempo**: Aumente o `end_time` para ver a semente atingir as bordas da grade.
- **Escala**: Tente uma grade de 50x50 para observar o padrão fractal que emerge com probabilidades baixas.

## Conclusão

O modelo de crescimento ilustra como regras probabilísticas simples podem gerar estruturas espaciais complexas e conectadas, servindo de base para modelos de percolação e difusão.
