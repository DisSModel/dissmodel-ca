# Modelo de Recozimento (Anneal)
### Autômato celular de votação majoritária com dissmodel

O modelo **Anneal** é uma variante de autômato celular baseada em votação majoritária que produz regiões suaves e arredondadas (semelhantes a bolhas ou manchas). É frequentemente utilizado para estudar a segregação espacial e a formação de fronteiras em sistemas biológicos ou físicos.

## Conceitos Fundamentais

Neste modelo, cada célula pode estar em um de dois estados:
- **L** (0): Estado "Esquerda" (Left)
- **R** (1): Estado "Direita" (Right)

A regra de transição depende da contagem de células no estado **L** na vizinhança de Moore (8 vizinhos + a própria célula):

| Células em L | Próximo Estado |
|:---:|:---:|
| 0 a 3 | **R** |
| 4 | **L** |
| 5 | **R** |
| 6 a 9 | **L** |

Esta regra específica cria um equilíbrio que favorece a formação de grandes áreas homogêneas com bordas minimizadas.

## Imports

Utilizaremos o `dissmodel` para a simulação e o `matplotlib` para definir as cores da visualização.

````python
from matplotlib.colors import ListedColormap

from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Anneal
````

## ⚠️ Ordem correta de inicialização

O `Environment` deve ser sempre criado **antes** de qualquer modelo (`Anneal`, `Map`). Os modelos se conectam ao ambiente ativo automaticamente no momento da instanciação.

```
Environment  →  Anneal  →  Map
     ↑            ↑         ↑
  primeiro     segundo   terceiro
```

````python
# 1. O Ambiente define o tempo da simulação
env = Environment(start_time=0, end_time=10)
````

## Construção do espaço/modelo

Criamos uma grade vetorial de 30x30 onde cada célula possui um atributo `state` inicializado como 0.

````python
gdf = vector_grid(dimension=(30, 30), resolution=1, attrs={"state": 0})
gdf.head()
````

## Inicialização

O método `initialize()` do modelo Anneal preenche a grade aleatoriamente com os estados L e R com probabilidade igual (50/50).

````python
# 2. O Modelo se conecta ao ambiente
model = Anneal(gdf=gdf)
model.initialize()
````

## Visualização

Configuramos um mapa para observar a evolução dos estados. Usaremos cores contrastantes para L e R.

````python
# 3. O Mapa também se conecta ao ambiente
cmap = ListedColormap(["#3498db", "#e74c3c"])  # Azul para L, Vermelho para R
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "#ecf0f1"},
)
````

## Execução da simulação

````python
env.run()
````
