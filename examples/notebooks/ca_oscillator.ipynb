# Modelo Oscilador (Oscillator)
### Sincronização e padrões de ondas em autômatos celulares

Este modelo, baseado no trabalho de Ermentrout & Edelstein-Keshet (1993), demonstra como células que oscilam individualmente podem se sincronizar espacialmente, formando frentes de onda e padrões dinâmicos complexos.

## Introdução

O modelo oscilador é usado para estudar fenômenos biológicos e físicos onde unidades individuais (como neurônios ou vaga-lumes) interagem para sincronizar seus ritmos. No contexto de autômatos celulares, isso resulta em ondas que se propagam pelo espaço e domínios de cores que pulsam em uníssono.

## Conceitos Fundamentais

### Estados
Cada célula possui um estado inteiro entre **0 e 15**. O estado representa a fase do ciclo de oscilação da célula.

### Regras de Transição
A mudança de estado depende se há algum vizinho no **estado 0**:
1. **Se houver vizinho no estado 0**: O próximo estado é ajustado por uma tabela de consulta (`AMOUNT`) que acelera ou retarda a oscilação para promover a sincronização.
2. **Caso contrário**: A célula simplesmente incrementa seu estado: $(estado + 1) \pmod{16}$.

**Vizinhança**: Moore (8 vizinhos).

## Imports

````python
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Oscillator
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  Oscillator  →  Map
```

````python
env = Environment(start_time=0, end_time=200)
````

## Construção do espaço/modelo

Criamos uma grade de 50x50 para permitir a visualização das frentes de onda.

````python
gdf = vector_grid(dimension=(50, 50), resolution=1, attrs={"state": 0})
````

## Inicialização

O modelo é inicializado com estados aleatórios (0-15) em cada célula.

````python
model = Oscillator(gdf=gdf)
model.initialize()
````

## Visualização

Como temos 16 estados, uma escala de cores (colormap) contínua ou ruidosa ajuda a perceber o movimento das ondas.

````python
Map(gdf=gdf, plot_params={"column": "state", "cmap": "twilight", "ec": "none"})
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Tempo de Simulação**: Este modelo demora alguns passos para sincronizar. Tente aumentar o `end_time` para 400 ou 500.
- **Escala Espacial**: Aumente a grade para 80x80. Padrões circulares ou espirais podem surgir dependendo da semente aleatória.
- **Cores**: Experimente mudar o `cmap` para `hsv` ou `magma` para visualizar as fases de formas diferentes.

## Conclusão

O modelo Oscilador é um exemplo fascinante de auto-organização, onde a desordem inicial (entropia alta) dá lugar a padrões espaciais ordenados e rítmicos através de interações locais.
