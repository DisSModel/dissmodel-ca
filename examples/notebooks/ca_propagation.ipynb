# Propagação Estocástica
### Um autômato celular de propagação espacial com probabilidade

Este modelo simula a disseminação de um fenômeno (como uma inovação tecnológica, uma doença ou um boato) em um espaço geográfico. A propagação ocorre através do contato entre vizinhos, modulada por uma probabilidade de sucesso.

## Introdução

Diferente de modelos puramente determinísticos, o modelo de Propagação introduz o elemento de incerteza. Ele ajuda a entender como a densidade inicial e a facilidade de transmissão influenciam a velocidade e a cobertura total da disseminação em um sistema espacial.

## Conceitos Fundamentais

- **Estados**:
    - `OFF` (Inativo/0): Célula suscetível a ser ativada.
    - `ON` (Ativo/1): Célula que já adquiriu o estado e pode transmiti-lo.
- **Vizinhança**: Utiliza **KNN (K-Nearest Neighbors)** com `k=4`, conectando cada célula aos seus 4 vizinhos mais próximos.
- **Regra Estocástica**:
    - Se uma célula já está `ON`, ela permanece `ON` para sempre (modelo cumulativo).
    - Se uma célula está `OFF` e tem **pelo menos um** vizinho `ON`, ela tem uma probabilidade $p$ (parâmetro `prob`) de se tornar `ON` no próximo passo.

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Propagation
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  Propagation  →  Map
```

````python
# 1. Criar o ambiente
env = Environment(start_time=0, end_time=30)
````

## Construção do espaço/modelo

Criamos uma grade de 30x30 células.

````python
gdf = vector_grid(dimension=(30, 30), resolution=1, attrs={"state": 0})
````

## Inicialização

Podemos configurar a probabilidade de propagação e a densidade inicial de células ativas.

````python
# 2. Instanciar o modelo com parâmetros personalizados
prop = Propagation(gdf=gdf)
prop.setup(prob=0.2, initial_density=0.05) # 20% de chance de propagar, 5% começam ativos
prop.initialize()
````

## Visualização

````python
# 3. Mapa com cores contrastantes
cmap = ListedColormap(["#f0f0f0", "#e74c3c"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "gray", "lw": 0.5},
)
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Efeito da Probabilidade**: O que acontece se `prob` for muito baixo (ex: 0.01) ou muito alto (ex: 0.9)? Observe a velocidade da mancha vermelha.
- **Densidade Inicial**: Se apenas uma única célula começar ativa (`initial_density` muito pequeno), a propagação ainda ocorre? 
- **Ponto Crítico**: Existe um valor de `prob` abaixo do qual a propagação tende a parar?
