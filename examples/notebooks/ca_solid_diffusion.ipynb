# Difusão em Sólidos (Mecanismo de Lacunas)
### Simulação de mistura atômica por movimento aleatório

Baseado no modelo clássico do NetLogo, este autômato simula como dois materiais sólidos diferentes se misturam no nível atômico através do deslocamento de lacunas (vacâncias) na rede cristalina.

## Introdução

Na ciência dos materiais, a difusão em sólidos não ocorre por fluxo livre, mas sim por átomos "pulando" para espaços vazios adjacentes. Este modelo captura essa dinâmica microscópica, mostrando como a entropia leva o sistema de um estado ordenado (separado) para um estado de mistura uniforme.

## Conceitos Fundamentais

- **Estados**:
    - `ÁTOMO TIPO 1` (0)
    - `ÁTOMO TIPO 2` (1)
    - `LACUNA` (2): Espaço vazio através do qual os átomos se movem.
- **Mecanismo de Difusão**:
    - A cada passo, cada lacuna seleciona aleatoriamente um vizinho que contenha um átomo.
    - A lacuna e o átomo trocam de lugar.
- **Vizinhança**: Moore (8 vizinhos).
- **Atualização Sequencial**: Para garantir a conservação de massa (átomos não podem ser criados ou destruídos), o modelo utiliza uma varredura aleatória sequencial em vez de uma atualização síncrona paralela.

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import SolidDiffusion
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  SolidDiffusion  →  Map
```

````python
# 1. Criar o ambiente com tempo longo para observar a mistura
env = Environment(start_time=0, end_time=400)
````

## Construção do espaço/modelo

Criamos uma grade de 31x31 (usamos ímpar para ter uma coluna central exata para as lacunas).

````python
dim = 31
gdf = vector_grid(dimension=(dim, dim), resolution=1, attrs={"state": 0})
````

## Inicialização

O método `initialize()` divide o cristal verticalmente: Átomos 1 à esquerda, Átomos 2 à direita e Lacunas no centro.

````python
# 2. Instanciar e inicializar
sd = SolidDiffusion(gdf=gdf, dim=31)
sd.initialize()
````

## Visualização

Usamos três cores: Vermelho (Átomo 1), Azul (Átomo 2) e Branco (Lacuna).

````python
# 3. Mapa
cmap = ListedColormap(["#e74c3c", "#3498db", "white"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "#bdc3c7", "lw": 0.1},
)
````

## Execução da simulação

````python
env.run()
````

## Conclusão

Embora o movimento de cada lacuna seja puramente aleatório (passeio aleatório), o resultado coletivo é um processo de difusão direcional que elimina o gradiente de concentração, demonstrando como leis macroscópicas emergem de comportamentos microscópicos simples.
