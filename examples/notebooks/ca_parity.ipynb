# Autômato de Paridade (Regra de Gilbert)
### Um autômato celular espacial com expansão em forma de diamante

O modelo de Paridade, baseado no trabalho de Nigel Gilbert, é um exemplo clássico de como regras de vizinhança simples podem gerar padrões geométricos complexos. Ele demonstra o crescimento de ondas em forma de diamante a partir de sementes iniciais.

## Introdução

Este modelo simula a propagação de estados em uma grade quadrada. A característica mais marcante do modelo de Paridade é a emergência de fractais e padrões auto-similares (como o triângulo de Sierpiński em certas condições) a partir de regras puramente locais. No contexto científico, ele serve para estudar processos de difusão e crescimento de fronteiras.

## Conceitos Fundamentais

- **Estados**: Cada célula pode estar em um de dois estados:
    - `ON` (Ativo/1)
    - `OFF` (Inativo/0)
- **Vizinhança**: Utiliza a vizinhança de **Von Neumann** (também conhecida como `Rook`), que considera os 4 vizinhos diretos: Norte, Sul, Leste e Oeste.
- **Regra de Transição**:

| Vizinhos Ativos | Próximo Estado |
|:---:|:---:|
| Exatamente 1 ou 3 | Ativo (ON) |
| 0, 2 ou 4 | Inativo (OFF) |

## Imports

Utilizamos o `Environment` para gerenciar o tempo, `vector_grid` para o espaço, e o modelo `Parity` pré-definido no pacote.

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Parity
````

## ⚠️ Ordem correta de inicialização

O `Environment` deve sempre ser criado **antes** do modelo e da visualização. Os componentes se conectam ao ambiente ativo automaticamente.

```
Environment  →  Parity  →  Map
```

````python
# 1. Criar o ambiente com tempo final suficiente para ver a expansão
env = Environment(start_time=0, end_time=100)
````

## Construção do espaço/modelo

Criamos uma grade de 50x50 células, onde todas começam no estado inativo (`state=0`).

````python
gdf = vector_grid(dimension=(50, 50), resolution=1, attrs={"state": 0})
gdf.head()
````

## Inicialização

O método `initialize()` posiciona duas sementes `ON` em posições estratégicas (x=10, y=25 e x=40, y=25) para demonstrar a interação das ondas.

````python
# 2. Instanciar o modelo e inicializar com as sementes
parity = Parity(gdf=gdf, dim=50)
parity.initialize()
````

## Visualização

Usamos um mapa com duas cores: branco para inativo e azul para ativo.

````python
# 3. Configurar o mapa
cmap = ListedColormap(["white", "royalblue"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "#eeeeee", "lw": 0.1},
)
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

Tente os seguintes experimentos:

- **Mudar o tempo**: Aumente `end_time` no `Environment` para ver o que acontece quando as ondas atingem as bordas da grade.
- **Grade maior**: Aumente a dimensão em `vector_grid` (ex: 100x100) e ajuste o parâmetro `dim` no modelo.
- **Novas sementes**: Tente alterar manualmente o estado de algumas células após a inicialização para ver como novos padrões surgem.

## Conclusão

O modelo de Paridade ilustra perfeitamente o conceito de **emergência**. Embora a regra local seja simples e baseada apenas na paridade da vizinhança, o resultado global é um padrão organizado e esteticamente belo que se propaga de forma rítmica pelo espaço.
