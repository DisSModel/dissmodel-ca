# Modelo de Meio Excitável (Excitable Medium)
### Propagação de ondas e padrões espirais com dissmodel

O modelo de **Meio Excitável** simula sistemas onde as unidades podem ser "estimuladas" e passam por um período de recuperação antes de poderem ser estimuladas novamente. Este fenômeno é observado em tecidos cardíacos, colônias de fungos ameboides (*Dictyostelium discoideum*) e reações químicas como a de Belousov-Zhabotinsky.

## Conceitos Fundamentais

As células ciclam por 6 estados (0 a 5):

1. **Repouso** (0): A célula está passiva, mas pode ser excitada.
2. **Excitado** (1): A célula foi estimulada por um vizinho.
3. **Refratário** (2-5): A célula está se recuperando e não pode ser excitada.

**Regras de Transição (Vizinhança de Rook/4 direções):**

- Se uma célula está em **Repouso** (0) e pelo menos um vizinho está **Excitado** (estado > 0), ela se torna **Excitada** (1).
- Se a célula já está **Excitada** ou em período **Refratário** (1-5), ela avança automaticamente para o próximo estado (1→2→3→4→5→0).

Este mecanismo impede que a onda retroceda, gerando frentes de propagação limpas e ondas circulares ou espirais.

## Imports

````python
from matplotlib.colors import LinearSegmentedColormap

from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Excitable
````

## ⚠️ Ordem correta de inicialização

O `Environment` deve ser sempre criado **antes** do modelo.

```
Environment  →  Excitable  →  Map
```

````python
env = Environment(start_time=0, end_time=100)
````

## Construção do espaço/modelo

Este modelo foi calibrado para uma grade de 50x50. Usamos `vector_grid` para criar o espaço celular.

````python
gdf = vector_grid(dimension=(50, 50), resolution=1, attrs={"state": 0})
gdf.head()
````

## Inicialização

O método `initialize()` coloca dois núcleos de excitação em posições estratégicas para gerar ondas que interagem entre si.

````python
model = Excitable(gdf=gdf)
model.initialize()
````

## Visualização

Usaremos um mapa de cores degradê para representar a transição entre os estados excitado e refratário.

````python
colors = ["#2c3e50", "#e74c3c", "#e67e22", "#f1c40f", "#27ae60", "#2980b9"]
cmap = LinearSegmentedColormap.from_list("excitable", colors, N=6)

Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "none"},
)
````

## Execução da simulação

````python
env.run()
````
