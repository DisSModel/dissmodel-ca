# Simulação de Queda de Neve
### Acúmulo e movimento de partículas em uma grade

Este autômato celular simula a dinâmica física simplificada de partículas de neve caindo e se acumulando no chão ou sobre outras partículas.

## Introdução

Ao contrário de modelos biológicos como o Jogo da Vida, este modelo foca no transporte de massa e na formação de estruturas por gravidade. Ele serve como base para entender sistemas onde partículas se movem em uma direção preferencial e interagem por bloqueio físico.

## Conceitos Fundamentais

- **Estados**:
    - `VAZIO` (0): Espaço livre.
    - `NEVE` (1): Partícula presente.
- **Dinâmica**:
    - **Geração**: Novos flocos de neve aparecem na linha do topo com uma probabilidade configurável.
    - **Queda**: Um floco de neve move-se para a célula de baixo se ela estiver vazia.
    - **Acúmulo**: Um floco para se atingir o fundo ou se a célula abaixo já estiver ocupada por neve.
- **Particularidade**: Este modelo calcula vizinhos diretamente através dos índices da grade (`row-col`), sem usar uma matriz de pesos fixa.

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Snow
````

## ⚠️ Ordem correta de inicialização

Para este modelo, é importante que o tempo total da simulação (`end_time`) seja maior que o dobro da dimensão da grade para que a neve tenha tempo de cair e acumular visivelmente.

```
Environment  →  Snow  →  Map
```

````python
dim = 20
# 1. Criar o ambiente (end_time > 2 * dim)
env = Environment(start_time=0, end_time=60)
````

## Construção do espaço/modelo

Criamos uma grade quadrada de 20x20.

````python
gdf = vector_grid(dimension=(dim, dim), resolution=1, attrs={"state": 0})
````

## Inicialização

Configuramos a probabilidade de um novo floco aparecer no topo a cada passo de tempo.

````python
# 2. Instanciar o modelo
snow = Snow(gdf=gdf, dim=20)
snow.setup(probability=0.3) # Alta probabilidade para ver o acúmulo rápido
snow.initialize()
````

## Visualização

Céu azul escuro (`#2c3e50`) e neve branca.

````python
# 3. Mapa
cmap = ListedColormap(["#2c3e50", "white"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "#34495e", "lw": 0.5},
)
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Nevasca vs. Flocos isolados**: Mude `probability` para `0.05` e depois para `1.0`. Como o padrão de acúmulo muda?
- **Duração**: Aumente o `end_time` no `Environment`. A neve chegará a cobrir toda a grade?
- **Janela de queda**: Observe no código-fonte que a neve para de cair algum tempo antes do fim da simulação para permitir que os últimos flocos pousem.
