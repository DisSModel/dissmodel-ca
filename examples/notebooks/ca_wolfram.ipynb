# Autômatos Elementares de Wolfram
### Regras 1D visualizadas em 2D (gerações por linha)

Stephen Wolfram explorou uma classe de autômatos celulares unidimensionais que, apesar de sua simplicidade extrema, exibem uma gama completa de comportamentos complexos, incluindo a computação universal.

## Introdução

Nesta implementação, o tempo é visualizado espacialmente: cada linha da grade representa o estado do autômato em um instante específico. A primeira linha (topo) é a condição inicial, e cada linha subsequente é calculada a partir da anterior.

## Conceitos Fundamentais

- **Espaço**: Uma linha de células (1D).
- **Vizinhança**: Para cada célula, considera-se ela mesma e seus dois vizinhos imediatos (esquerda e direita).
- **Regras de Wolfram**: Existem $2^8 = 256$ regras possíveis. Cada regra define o estado da célula central com base nas 8 combinações possíveis dos 3 vizinhos.
- **Visualização**: 
    - Eixo X: Espaço (as células do autômato).
    - Eixo Y: Tempo (cada linha é uma nova geração).

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import Wolfram
````

## ⚠️ Ordem correta de inicialização

A altura da grade (`ydim`) deve ser igual ao número de passos (`end_time + 1`) para que todas as gerações sejam exibidas.

```
Environment  →  Wolfram  →  Map
```

````python
final_time = 50
x_dim = 101 # Largura ímpar para ter um centro exato

# 1. Criar o ambiente
env = Environment(start_time=0, end_time=final_time)
````

## Construção do espaço/modelo

Criamos uma grade retangular onde as colunas são o espaço e as linhas são o tempo.

````python
gdf = vector_grid(dimension=(x_dim, final_time + 1), resolution=1, attrs={"state": 0})
````

## Inicialização

Podemos escolher qualquer regra de 0 a 255. A Regra 90 é famosa por gerar o Triângulo de Sierpiński. A inicialização coloca um único ponto ativo no centro da primeira linha.

````python
# 2. Instanciar o modelo
wolf = Wolfram(gdf=gdf, rule_number=90, final_time=final_time, xdim=101)
wolf.initialize()
````

## Visualização

````python
# 3. Mapa (Preto e Branco)
cmap = ListedColormap(["white", "black"])
Map(
    gdf=gdf,
    plot_params={"column": "state", "cmap": cmap, "ec": "none"},
)
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

A classificação de Wolfram divide as regras em 4 classes. Experimente:

- **Classe 1 (Estável)**: Regra 0, Regra 32.
- **Classe 2 (Periódica)**: Regra 4, Regra 108.
- **Classe 3 (Caótica/Fractal)**: Regra 30, Regra 90, Regra 150.
- **Classe 4 (Complexa/Universal)**: Regra 110 (capaz de computação universal!).

Basta alterar o `rule_number` e rodar as células novamente.

## Conclusão

Os autômatos de Wolfram mostram que a complexidade não requer regras complexas. Sistemas extremamente simples podem produzir comportamentos imprevisíveis e ricos, um princípio fundamental da teoria dos sistemas complexos.
