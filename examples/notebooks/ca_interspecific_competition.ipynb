# Competição Interespecífica
### Dinâmica espacial de cinco espécies de gramíneas

Este modelo simula a competição por espaço entre cinco espécies de gramíneas, baseado no estudo clássico de Silvertown et al. (1992). Ele demonstra como a disposição inicial das espécies afeta o resultado final da competição.

## Introdução

Na ecologia de comunidades, a competição por espaço é um fator determinante para a biodiversidade. Este modelo utiliza probabilidades de invasão determinadas empiricamente para simular como uma espécie pode substituir outra em uma célula da grade, dependendo de quem são seus vizinhos.

## Conceitos Fundamentais

### Espécies (Estados)
O modelo inclui cinco espécies:
0. *Lolium perenne*
1. *Agrostis stolonifera*
2. *Holcus lanatus*
3. *Poa trivialis*
4. *Cynosurus cristatus*

### Regras de Transição
A probabilidade de uma espécie $A$ invadir uma célula ocupada pela espécie $B$ depende de:
1. A fração de vizinhos da espécie $A$.
2. A taxa de invasão experimental de $A$ sobre $B$.

**Vizinhança**: Von Neumann (4 vizinhos - Rook).

## Imports

````python
from matplotlib.colors import ListedColormap
from dissmodel.core import Environment
from dissmodel.geo import vector_grid
from dissmodel.visualization.map import Map
from dissmodel_ca.models import InterspecificCompetition
````

## ⚠️ Ordem correta de inicialização

```
Environment  →  InterspecificCompetition  →  Map
```

````python
env = Environment(start_time=0, end_time=100)
````

## Construção do espaço/modelo

O modelo original utiliza uma grade de 40x40 para respeitar as proporções das faixas iniciais.

````python
gdf = vector_grid(dimension=(40, 40), resolution=1, attrs={"state": 0})
````

## Inicialização

Podemos escolher diferentes arranjos iniciais: `ModelA`, `ModelB`, `ModelC` (faixas horizontais) ou `Random` (aleatório).

````python
model = InterspecificCompetition(gdf=gdf, displacement="ModelA")
model.initialize()
````

## Visualização

````python
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]
cmap = ListedColormap(colors)
Map(gdf=gdf, plot_params={"column": "state", "cmap": cmap, "ec": "none"})
````

## Execução da simulação

````python
env.run()
````

## Exploração de parâmetros / Experimentos guiados

- **Arranjo Inicial**: Altere `displacement` para `ModelB`, `ModelC` ou `Random`. Como a ordem inicial das faixas afeta qual espécie domina no final?
- **Tempo**: Aumente o `end_time` para 500 ou mais para ver se a comunidade atinge um estado estável ou se algumas espécies são extintas localmente.

## Conclusão

Este modelo demonstra que, em ecologia, não apenas as habilidades competitivas importam, mas também a configuração espacial e a vizinhança imediata das populações.
